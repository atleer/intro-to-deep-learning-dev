Define an empty class named `MyClass`
```python
class MyClass:
    pass
```

Define a class named `MyClass` that has a function named `myclass_function`.

```python
class MyClass:
    def myclass_function(self):
        print("This is a function in MyClass")
```

Create an instance of `Myclass`:

```python
my_instance = MyClass()
```

Call `MyClass`'s function `myclass_function`:

```python
my_instance.myclass_function()
```

Define a class named `MySubClass` that inherits from MyClass:

```python
class SubClass(MyClass):
    pass
```

Create an instance of `SubClass` and call the function `myclass_function`, which is inherited from `MyClass`:

```python
my_new_instance = SubClass()

my_new_instance.myclass_function()
```
