## Section: Functions Have Expectation on their Inputs.  Type Annotations can communicatie this intent

What is the motivation for this session? What questions does it answer?

- How can we make it easier to work with models?
  - Through good interfacing.
- How do we make a good interface in general?
  - Through OOP and type annotation.
- How do we make an interface for deep learning?
  - By leveraging PyTorch's neural network (nn) module
    - Without it, you'd have to track parameters for each layer manually instead of getting all parameters together with `model.parameters()` - e.g, for optimizers, you can write `optimizer = torch.optim.Adam(model.parameters())` instead of `optimizer = torch.optim.Adam(list(layer1.parameters())+list(layer2.parameters()+...etc))`
    - Switching device (model.to(device='cpu') or model.to(device='cuda')) much easier.
    - (Maybe less important, but you get a printing of the model with all its layers.)
  - Through encapsulation.

## Section 1: Models Expect a Certain Input. Type Annotations Can Communicate This Expectation to The User

| Code | Description |
| :-- | :-- |
| `x: int` | `x` is an integer. |
| `x: int = 5` | `x` is an integer with the value 5. |
| `x: str` | `x` is a string. |
| `x: float` | `x` is a float. |
| `x: bool` | `x` is a boolean. |
| `x: list` | `x` is a list. |
| `x: list[float]` | `x` is a list containing elements that are floats. |
| `x: torch.Tensor` | `x` is a tensor. |
| `def fun(x: int)` | The argument `x` in `fun` is an integer. |
| `def fun(x: list)` | The argument `x` in `fun` is a list. |
| `def fun(x: np.ndarray)` | The argument `x` in `fun` is a numpy array. |
| `def fun() -> torch.Tensor:` | The output of the function `fun` is a tensor. |
| `def fun(x: np.ndarray) -> torch.Tensor:` | The argument `x` is a float and the is a tensor. |

The way the `DairyProducts` and `Expired` classes are defined now, different instantiations of the classes will always have the same number of milk cartons and yoghurts. But of course, different stores usually have different quantities of those products. To make it possible to create instances with different values for the attributes, we need to define the class with an `__init__`  function.

**Example**: Create a new class named `DairyProducts` with an `__init__` function that initializes the attributes `num_milk_cartons` and `num_yoghurts`.

**Example**: Create a class named `Model` that inherits from the `nn.Module` and is initialized with a three layers. The first layer has 11 input features and 16 output features. The second layer has 16 input features and 16 output feature. The last layer has 16 input features and 1 output feature. Add a forward method to the model.

````python
class Model(nn.Module):
    def __init__(self):
        super().__init__()
        self.layer1 = nn.Linear(1,16)
        self.layer2 = nn.Linear(16,16)
        self.layer3 = nn.Linear(16,1)

    def forward(self, x):
        out1 = self.layer1(x)
        out2 = self.layer2(out1)
        out3 = self.layer3(out2)
        return out3
    
x = torch.linspace(0,10,11).unsqueeze(1)# torch.tensor(5.0)
print(x.shape)
Model().forward(x)
````
