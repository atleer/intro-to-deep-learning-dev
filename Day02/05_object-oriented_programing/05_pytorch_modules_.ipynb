## Section 1: Models Expect a Certain Input. Type Annotations Can Communicate This Intent to The User

| Code | Description |
| :-- | :-- |
| `def fun(x: integer)` | The argument `x` in `fun` is an integer. |
| `def fun(x: string)` | The argument `x` in `fun` is a float. |
| `def fun(x: float)` | The argument `x` in `fun` is a float. |
| `def fun(x: Tensor)` | The argument `x` in `fun` is a tensor. |
| `def fun(x: bool)` | The argument `x` in `fun` is a boolean. |
| `def fun(x: list[])` | The argument `x` in `fun` is a list. |
| `def fun(): -> tensor` | The output of the function `fun` is a tensor. |

## Section: Customizing Your Python Classes using Inhertiance in Python

```python
class A:

    def arbitrary(self, new) -> int:
        print('override me!')
        return 3


def run(x: A):
    return x.arbitrary(1) * 2



class B(A):

    # Where to store your data
    def __init__(self): 
        self.stored = 3


    # function that can run on stored and new data
    def arbitrary(self, new): 
        return self.stored + new
    



b = B()
b.arbitrary(4)
run(b)
```
