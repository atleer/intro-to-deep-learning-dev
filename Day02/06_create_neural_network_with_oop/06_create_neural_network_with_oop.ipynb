## Section 1: Make a Basic Neural Network Class

| Code | Description |
| --- | --- |
| `def __init__(self):` | Initializes the class. |
| `super(ClassName,self).__init__()` | Makes the class `ClassName` inherit from the parent class. |
| `nn.Sequential(`<br>&nbsp;&nbsp;&nbsp;&nbsp;`nn.Linear(1,10),`<br>&nbsp;&nbsp;&nbsp;&nbsp;` nn.ReLU(),`<br>&nbsp;&nbsp;&nbsp;` nn.Linear(10,1)`<br>`)` | Creates a container where layers and activation functions are stacked in a sequential order. This ensures that the input is passed through the network in that order.|
| `def forward(self, x):`<br>&nbsp;&nbsp;&nbsp;&nbsp;`output = self.layers(x)`<br>&nbsp;&nbsp;&nbsp;&nbsp;`return output` | Create forward function that is used to pass input through network defined in a class. |
| ``net = ClassName()`` | Create an instance of the class `ClassName` and assigns it to the variable `net`. |

**Exercise**: Create a class called TwoLayerNeuralNet. It needs to inherit from nn.Module and consists of a two layers with a rectified linear unit (``nn.ReLU()``) between them. The first layer takes 13 input features and 32 output features. The second layer takes 32 input features and 13 output features. Display it.

**Hint**: Use `nn.Sequential(...)` to ensure that input is passed through the network in the right order.

`````python
class SimpleNeuralNet1(nn.Module):
    def __init__(self,):
        super(SimpleNeuralNet1, self).__init__()

        self.layers = nn.Sequential(
            nn.Linear(13, 16),
            nn.ReLU(),
            nn.Linear(16,13)
        )
`````

`````python
class SimpleNeuralNet2(nn.Module):
    def __init__(self,):
        super(SimpleNeuralNet2, self).__init__()

        self.layers = nn.Sequential(
            nn.Linear(13, 16),
            nn.ReLU(),
            nn.Linear(16,13)
        )

    def forward(self, x):
        output = self.layers(x)
        return output
`````
