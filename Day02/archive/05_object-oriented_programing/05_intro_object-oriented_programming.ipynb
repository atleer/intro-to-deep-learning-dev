#### Adding Methods

````python
class Student:
    def __init__(self):
        self.name = 'Jack'
        self.course = 'Astronomy101'
        self.grade = None

    def set_grade(self, grade):
        self.grade = grade

stud = Student()
print(stud.grade)
````

````python
stud.set_grade(2.0)
print(stud.grade)
````

## Section 2: Inheritance

To reduce code duplication, it's possible to make a class that automatically has all the methods of another class:

```python
class Material:
    def __init__(self, weight, strength): ...
    def is_heavier_than(self, other): ...
    def is_stronger_than(self, other): ...

aluminum = Material(weight=3.2, strength=1.3)

class Paper(Material):
    def tear(self): ...
    def burn(self): ...


sheet = Paper(weight=1.1, strength=0.1)
sheet.is_heavier_than(aluminum)
sheet.tear
```

**Note:** Any methods that the new class adds will *overwrite* the methods of the class it inherits from.

````python
class Person:
    def __init__(self):
        self.name = None
        self.age = None
        self.nationality = 'German'

    def change_nationality(self, nationality):
        self.nationality = nationality
````

````python
class Student(Person):
    def __init__(self):
        super().__init__()

        self.course = None
````

````python
stud = Student()
print(stud.nationality)
````

````python
stud.change_nationality('French')
print(stud.nationality)
````
