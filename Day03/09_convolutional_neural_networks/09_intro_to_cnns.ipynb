```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
```

## Section 1: Why Convolution? Neighbouring Data is Often Similar -> We Can Reduce # of Parameters Through Sharing and Save Computation

| Code | Description |
| --- | --- |
| `nn.Conv2d(in_channels, out_channels, kernel_size)` | Creates a 2D convolution to be applied to the signal. |

```python
a = np.zeros(100)
a[30:70] = 1

x = np.linspace(0,10,100)
#mu, sigma = np.median(x), 1
#kernel = 1*np.exp(-(x-mu)**2/(2*sigma**2))
kernel = np.zeros(100)
kernel[45:55] = 1
kernel_norm_factor = kernel.sum()
kernel /= kernel_norm_factor

result = np.zeros_like(a)

def do_and_plot_convolution(idx):
    
    shifted_kernel = np.roll(kernel, idx-len(kernel) // 2)
    result[idx] = (a*shifted_kernel).sum()
    
    fig, axes = plt.subplots(nrows = 2, figsize = (5,5))
    
    axes[0].plot(a, color = 'b', label = 'Original vector')
    axes[0].plot(shifted_kernel*kernel_norm_factor, color = 'g', label = f'(Non-normalized) Kernel at index {idx}')
    axes[1].plot(np.arange(len(a)), result)
    axes[1].set_ylabel('Result of Convolution')
    fig.legend(bbox_to_anchor = (1.6,0.8))
    plt.tight_layout()
    plt.show()

slider = widgets.IntSlider(min = 25, max = len(a)-1, step = 1, description = 'index')
widgets.interact(do_and_plot_convolution, idx=slider);
```
