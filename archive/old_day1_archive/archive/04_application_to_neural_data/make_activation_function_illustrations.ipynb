```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
fig_dir = Path('img')
fig_dir.mkdir(exist_ok=True, parents=True)
```

ReLU

```python
def ReLU(x):
    y = np.copy(x)
    y[y <= 0] = 0
    return y
```

```python
x = np.linspace(-3,3,100)
out = ReLU(x)
plt.figure(figsize=(4,3))
plt.plot(x,out)
plt.xlabel('x (input)')
plt.ylabel('y (output)')
plt.title('Illustration ReLU\nActivation Function')
plt.tight_layout()
plt.savefig(fig_dir / 'illustration_relu.png')
```

Hyperbolic tangent

```python
def tanh(x):
    return (np.exp(x)-np.exp(-x))/(np.exp(x)+np.exp(-x))
```

```python
x = np.linspace(-5,5,100)
out = tanh(x)
plt.figure(figsize=(4,3))
plt.plot(x, out)
plt.xlabel('x (input)')
plt.ylabel('y (output)')
plt.title('Illustration Hyperbolic Tangent\nActivation Function')
plt.tight_layout()
plt.savefig(fig_dir / 'illustration_tanh.png')
```

Sigmoid

```python
def sigmoid(x):
    return 1/(1+np.exp(-x))
```

```python
x = np.linspace(-5,5,100)
out = sigmoid(x)
plt.figure(figsize=(4,3))
plt.plot(x, out)
plt.xlabel('x (input)')
plt.ylabel('y (output)')
plt.title('Illustration Sigmoid\nActivation Function')
plt.tight_layout()
plt.savefig(fig_dir / 'illustration_sigmoid.png')
```
