## Section 1: Machine Learning is iterative guessing.  Iteration can be done with For Loops.

When you train a neural network, you're making the model guess what the output should be over and over again. Between each guess, the model parameters are updated until the guess is good enough. In other words, it's an iterative improvement of the model. This iteration can be programmed with a for loop. A for loop lets us perform an operation a predefined number of times and is an incredibly useful programming concept. In this section, you will get som practice writing for loops.

| Code | Description |
| --- | --- |
| `for letter in ["a", "b", "c"]:` <br> &nbsp;&nbsp;&nbsp;&nbsp; `print(letter)` | Print every `letter` in the list `["a", "b", "c"]`|
| `for i in range(5):` <br> &nbsp;&nbsp;&nbsp;&nbsp; `print(i)` | Print every number `i` between 0 and 5 (not including 5)|

| `for x_i in x:` <br> &nbsp;&nbsp;&nbsp;&nbsp; `for y_i in y`: <br> &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; `print(x_i, y_i)` | Print every element `x_i` and `y_i` in the collections `x` and `y`|

## Section 2: Machine Learning is Adjusting Guesses Based on Error. 

| Code | Description |
| --- | --- |
| `param = param - learning_rate * gradient * param` | Update the parameter `param`. |
| `z**2` | Calculate the square of z. |
| `y_pred = a * x` | Make a prediction using parameter `a` and input `x`. |
| `(y_pred - y_obs)**2` | Calculate the squared error (loss) between prediction and observation. |
| `grad_a = -2 * (y_obs - y_pred) * x` | Calculate the gradient of the loss with respect to parameter `a`. |
| `a = a - learning_rate * grad_a` | Update parameter `a` using the gradient and learning rate. |
| `learning_rate = some_num` | Define the learning rate hyperparameter, which controls the step size of gradient descent. |

Let's suppose we have some recorded data that we know can be described by a function of the form:

$$y = a \times x$$

However, we don't know the value of `a`. We could of course use one of the standard linear regression functions that many python libraries offer to do this, but let's do it with machine learning instead. And we'll do it manually to get a better intuition for what is actually going on under the hood in machine learning.

To keep it simple, say we have a single observation  `y_obs = 10.0` recorded at `x = 2.0`. In the cell below, we start with an initial guess `a = 1.0`. Run the cell to see how close the prediction `y_pred` is to the observed `y_obs` with this initial guess for `a`.

````python
# Data
x = 2.0
y_obs = 10.0

# Parameter
a = 1.0 # initial guess

y_pred = a * x
print(f"Intial guess a = {a}; observed y = {y_obs}, predicted y = {y_pred}, difference = {y_obs-y_pred}")
````

The predicted `y_pred` wasn't very close to our observed `y_obs`. The deviation between the prediction and the observed data is referred to as the "loss" in the machine learning context. A common way to quantify the loss is to calculate the squared difference between the observed and the predicted value.

**Exercise**: Calculate the squared difference between `y_pred` and `y_obs`.

````python
(y_pred - y_obs)**2
````

We can use the loss to find out how to do better with our next guess. That's where the **gradient** comes in. *The gradient tells us how much the loss changes if we change a parameter in our model.* In our model here, it tells us how much the deviation between our prediction and the observed data `y_obs` will change if we adjust `a`. In mathematical terms, the gradient is the derivative of the loss with respect to the parameter.

The math below shows how the gradient for `a` is derived if you want to know where the final equation comes from, but you can do the next exercise without going through the derivation. We start with definition of the gradient: the change in loss ($`\Delta loss`$) per change in the parameter ($`\Delta a`$).

$$grad_a = \frac{\Delta loss}{\Delta a}$$

When the change in `a` is small, this can be written as a differential equation:

$$grad_a = \frac{\partial{loss}}{\partial{a}}$$

we insert $loss = (y_{obs}-y_{pred})^2$

$$ grad_a = \frac{\partial{(y_{obs}-y_{pred})}}{\partial{a}}$$

we insert $y_{pred} = a \times x$

$$ grad_a = \frac{\partial{(y_{obs}-a \times x)}}{\partial{a}}$$

we calculate the derivative

$$ grad_a = -2(y_{obs}-a \times x) \times x$$

we write $`y_{obs}`$ instead of $`a \times x`$ in the final equation

$$ grad_a = -2(y_{obs}-y_{pred}) \times x$$

**Exercise**: Write the **last** equation above in code.


````python
grad_a = -2*(y_obs-y_pred)*x
grad_a
````

We'll now use the gradient we calculated to update the parameter `a` before our next guess.

**Exercise**: Calculate a new parameter value `a_new` using the current value of `a` and the gradient calculated above.

**Hint**: Check the reference table to see how parameters can be updated using the gradient.

````python
a = a-grad_a*a
a
````

Now we're ready to make our next guess. 

**Exercise**: Calculate `y_pred` again but use the new parameter value `a_new` instead of the original `a`.

**Note**: This guess will actually be worse than the initial guess, so don't worry, that's to be expected.

````python
y_pred = a * x
y_pred
````

There is an important *hyperparameter* (different from a regular parameter) called the **learning rate**, which determines how we learn in machine learning. Without an appropriate learning rate, the updated parameter values won't give us better predictions. The lack of an appropriate learning rate is the reason why our second guess was worse than the first guess above. The learning rate is multiplied with the gradient and the current parameter value in the parameter update.

**Exercise**: Calculate the updated parameter value `a` again, but this time, multiply the gradient and the current parameter value with the `learning_rate` defined in the cell below.

````python
a = a-learning_rate*grad_a*a
a
````

## Section 3: `torch.tensor()` has Autograd directly implemented, so it can automatically compute deriviates. 

The great thing about PyTorch is that it takes care of the gradient calculation and parameter updates for you, so you don't need to do it manually like you did above. This is particularly useful when you have a big model with a lot of parameters and layers that need to be kept track of and updated. 

```python
x = torch.tensor([2.], requires_grad=True)
y = x ** 2
y.backward()
x.grad  # tensor([7.])
```


| Code | Description |
| :-- | :-- |
| `x = torch.tensor([content_of_tensor], requires_grad=True)` | Create tensor and specify that gradients should be tracked. |
| `x.requires_grad` | Check if gradient tracking is on. |
| `y.backward()` | Do backpropagation on `y`. |
| `x.detach()` | Stop gradient tracking by detaching the tensor from the graph. |
| `with x.no_grad():` | Pause gradient tracking temporarily|
| `x.zero_()` | Reset the gradient. |

Below, we find the `a` that fits the data in the equation $y = a * x$, like you did in section 2. But this time, you'll use autograd instead.

**Exercise**: Run the first cell below to initialize the data variables `x`, `a`, and `y_obs`. Then put the code in the **second** cell in a for loop that stops after 5 steps.

````python
# initialize parameters
a = torch.tensor([1.0], requires_grad=True) # parameter, gradient needed
learning_rate = 0.1

# data
x = torch.tensor([2.0]) # input, no gradient needed
y_obs = torch.tensor([10.0]) # output, no gradient needed
````

````python
# forward pass
y_pred = a * x

# calculate the loss
loss = (y_pred - y_obs)**2

print(f"step {step}: a={a.item():.3f}, y_obs={y_obs.item():.2f}, y_pred={y_pred.item():.2f}, loss={loss.item():.3f}")

# backward pass — autograd computes dl/da for us
loss.backward()

# update parameter
with torch.no_grad():
    a -= learning_rate * a.grad

# reset gradient for next step
a.grad.zero_()
````

````python
for step in range(5):

    # forward pass
    y_pred = a * x

    # calculate the loss
    loss = (y_pred - y_obs)**2

    print(f"step {step}: a={a.item():.3f}, y_pred={y_pred.item():.2f}, loss={loss.item():.3f}")

    # backward pass — autograd computes dl/da for us
    loss.backward()

    # update parameter
    with torch.no_grad():
        a -= learning_rate * a.grad

    # reset gradient for next step
    a.grad.zero_()
````

````python
# data
x = 2.0
y_obs = 10.0
def train_without_autograd():
    # initialize parameters
    a = 1.0
    learning_rate = 0.1
    for step in range(5):
        # Forward pass
        y_pred = a * x

        # calculate loss
        loss = (y_pred - y_obs)**2

        # Compute gradient
        grad_a = 2 * (y_pred - y_obs) * x

        # Update the parameter
        a = a - learning_rate * grad_a
    
time_without = timeit.timeit(train_without_autograd, number = 10)
print(f"Without autograd: {time_without:.5f} seconds")
````

````python
# data
x = torch.tensor([2.0]) # input, no gradient needed
y_obs = torch.tensor([10.0], requires_grad=True) # output, gradient needed

def train_with_autograd():
    # initialize parameters
    a = torch.tensor([1.0], requires_grad=True) # parameter, gradient needed
    learning_rate = 0.1
    for step in range(5):
        # Forward pass
        y_pred = a * x
        loss = (y_pred - y_obs)**2

        # Backward pass — autograd computes dl/da for us
        loss.backward()

        # Update parameter
        with torch.no_grad():
            a -= learning_rate * a.grad

        # Reset gradient for next step
        a.grad.zero_()

time_with = timeit.timeit(train_with_autograd, number=10)
print(f"With autograd: {time_with:.5f} seconds")
````
