## Section 2: Machine Learning is Adjusting Guesses Based on Loss (Error) 

To improve the model between each guess, we need to know how close the model was to getting the right answer; we need to know the difference between the model's prediction and the true value. This is referred to as the "loss", and the machine learns by minimizing this loss until it gets the model's prediction as close to the true value as it can. In this section, we'll learn how to calculate the loss and explore it's role in updating parameters between iterations in machine learning.

| Code | Description |
| --- | --- |
| `param = param - learning_rate * gradient_param` | Update the parameter `param`. |
| `z**2` | Calculate the square of z. |
| `y_pred = a * x` | Make a prediction using parameter `a` and input `x`. |
| `(y_pred - y_obs)**2` | Calculate the squared error (loss) between prediction and observation. |
| `grad_a = -2 * (y_obs - y_pred) * x` | Calculate the gradient of the loss with respect to parameter `a`. |
| `a = a - learning_rate * grad_a` | Update parameter `a` using the gradient and learning rate. |
| `learning_rate = some_num` | Define the learning rate hyperparameter, which controls the step size of gradient descent. |

We want to quantify the relationship between the firing rate of a neuron in visual cortex and the brightness of a stimulus. Let's suppose that it's described by a simple linear equation where `x` is stimulus strength and `y` is the firing rate of the neuron:

$$y = a \times x$$

But we don't know the parameter `a`, so we start by just taking a guess. We guess that `a = 2` and predict the firing rate for stimulus strength `x = 2` using the equation above. That gives us the predicted firing rate `y_pred = 4` Hz. Now, we compare this prediction to our recorded data to see how good it was.

**Demo**: We can write the iterative guessing above using a loop.

````python
x = 2
y_obs = 10 # Hz

for a in [3,2,4,5]:
    y_pred = a * x # Hz
    loss = np.mean((y_pred-y_obs)**2)

    print(f'a: {a}, y_obs: {y_obs}, y_pred: {y_pred}, loss: {loss}')
````

Above, we manually changed the direction of the guess - whether to decrease or increase `a` - based on the loss. But the whole point of machine learning is that the machine should do this automatically.

Gradient descent offers a way to automatically determine which direction to change the parameters in order to minimize the loss. It provides an update rule for the next parameter guess. The update rule for the parameter `a` above is expressed as follows:

$$ a_{next} = a_{current} - \eta \times grad_a$$

where $\eta$ is the learning rate (how much your parameters should change from one guess to the next) and $grad_a$ is the gradient of `a`. The gradient tells us how much the loss will change for a change in the parameter, and the equation for the gradient of parameter `a` can be written as:

$$ grad_a = \frac{\partial{loss}}{\partial{a}} = -2 \times (y_{obs} - y_{pred}) \times x$$

This update of parameter values by calculating the gradient of the loss is what's referred to as "backpropagation".

## Section 4: Different Problems Require Different Kinds of Loss Functions

So far, we've used mean squared error (MSE) as our loss function. MSE works well for regression problems, where we're trying to predict a continuous value like a firing rate or a price. But not all machine learning problems are regression problems. In classification problems, we're trying to predict which category something belongs to - for example, whether an image shows a cat or a dog, or which type of stimulus was presented to the animal in an experiment. For these problems, MSE doesn't work well because we're not measuring distance from a target value, we're measuring how confident the model is in its class prediction. PyTorch provides different loss functions for different types of problems:

- **MSE Loss** (`nn.MSELoss()`): For regression - predicting continuous values
- **Binary Cross-Entropy** (`nn.BCEWithLogitsLoss()`): For binary classification - two classes
- **Cross-Entropy** (`nn.CrossEntropyLoss()`): For multiclass classification - more than two classes

In all cases, the loss measures how wrong the model's predictions are. Lower loss means better predictions, and training a neural network means finding parameters that minimize the loss.

| Code | Description |
| --- | --- |
| `nn.MSELoss()` | Creates a mean squared error (MSE) loss function. Used for regression problems. |
| `nn.BCEWithLogitsLoss()` | Creates a binary cross-entropy loss function. Used for binary classification (two classes). |
| `nn.CrossEntropyLoss()` | Creates a cross-entropy loss function. Used for multiclass classification (more than two classes). |
| `loss_fn(y_pred, y_true)` | Calculates the loss between predictions `y_pred` and true values `y_true`. |
| `loss_fn(y_pred, y_true).item()` | Extracts the loss value as a `float` from the tensor. |

Below, we find the `a` that fits the data in the equation $y = a * x$, like you did in section 2. But this time, you'll use autograd instead.

````python
# initialize parameters
a = torch.tensor([1.0], requires_grad=True) # parameter, gradient needed
learning_rate = 0.1

# data
x = torch.tensor([2.0]) # input, no gradient needed
y_obs = torch.tensor([10.0]) # output, no gradient needed
````

````python
# forward pass
y_pred = a * x

# calculate the loss
loss = (y_pred - y_obs)**2

print(f"step {step}: a={a.item():.3f}, y_obs={y_obs.item():.2f}, y_pred={y_pred.item():.2f}, loss={loss.item():.3f}")

# backward pass — autograd computes dl/da for us
loss.backward()

# update parameter
with torch.no_grad():
    a -= learning_rate * a.grad

# reset gradient for next step
a.grad.zero_()
````

````python
for step in range(5):

    # forward pass
    y_pred = a * x

    # calculate the loss
    loss = (y_pred - y_obs)**2

    print(f"step {step}: a={a.item():.3f}, y_pred={y_pred.item():.2f}, loss={loss.item():.3f}")

    # backward pass — autograd computes dl/da for us
    loss.backward()

    # update parameter
    with torch.no_grad():
        a -= learning_rate * a.grad

    # reset gradient for next step
    a.grad.zero_()
````

#### Derivation of Gradient Equation

We can use the loss to find out how to do better with our next guess. That's where the **gradient** comes in. *The gradient tells us how much the loss changes if we change a parameter in our model.* In our model here, it tells us how much the deviation between our prediction and the observed data `y_obs` will change if we adjust `a`. In mathematical terms, the gradient is the derivative of the loss with respect to the parameter.

The math below shows how the gradient for `a` is derived if you want to know where the final equation comes from, but you can do the next exercise without going through the derivation. We start with definition of the gradient: the change in loss ($`\Delta loss`$) per change in the parameter ($`\Delta a`$).

$$grad_a = \frac{\Delta loss}{\Delta a}$$

When the change in `a` is small, this can be written as a differential equation:

$$grad_a = \frac{\partial{loss}}{\partial{a}}$$

we insert $loss = (y_{obs}-y_{pred})^2$

$$ grad_a = \frac{\partial{(y_{obs}-y_{pred})}}{\partial{a}}$$

we insert $y_{pred} = a \times x$

$$ grad_a = \frac{\partial{(y_{obs}-a \times x)}}{\partial{a}}$$

we calculate the derivative

$$ grad_a = -2(y_{obs}-a \times x) \times x$$

we write $`y_{obs}`$ instead of $`a \times x`$ in the final equation

$$ grad_a = -2(y_{obs}-y_{pred}) \times x$$

Using more mathematical notation with the gradient expressed with $\nabla$ and $\mathcal{L}$ for the loss, we get:

$$ \nabla_a(\mathcal{L}) = -2 \times (y_{obs}-y_{pred}) \times x$$

````python
start = time.perf_counter()

for step in range(100000):
    # Forward pass
    y_pred = a * x

    # calculate loss
    loss = (y_pred - y_obs)**2

    # Compute gradient
    grad_a = 2 * (y_pred - y_obs) * x

    # Update the parameter
    a = a - learning_rate * grad_a

end = time.perf_counter()

print(f"Time without autograd: {end-start:.4f} seconds")
````

````python
# initialize parameters
a = torch.tensor([1.0], requires_grad=True) # parameter, gradient needed
learning_rate = 0.1

# data
x = torch.tensor([2.0]) # input, no gradient needed
y_obs = torch.tensor([10.0], requires_grad=True) # output, gradient needed
````

````python
start = time.perf_counter()
for step in range(5):
    # Forward pass
    y_pred = a * x
    loss = (y_pred - y_obs)**2

    # Backward pass — autograd computes dl/da for us
    loss.backward()

    # Update parameter
    with torch.no_grad():
        a -= learning_rate * a.grad

    # Reset gradient for next step
    a.grad.zero_()
end = time.perf_counter()

print(f"Time with autograd: {end-start:.4f} seconds")
````

````python
# data
x = torch.tensor([2.0]) # input, no gradient needed
y_obs = torch.tensor([10.0], requires_grad=True) # output, gradient needed

def train_with_autograd():
    # initialize parameters
    a = torch.tensor([1.0], requires_grad=True) # parameter, gradient needed
    learning_rate = 0.1
    for step in range(5):
        # Forward pass
        y_pred = a * x
        loss = (y_pred - y_obs)**2

        # Backward pass — autograd computes dl/da for us
        loss.backward()

        # Update parameter
        with torch.no_grad():
            a -= learning_rate * a.grad

        # Reset gradient for next step
        a.grad.zero_()

time_with = timeit.timeit(train_with_autograd, number=10)
print(f"With autograd: {time_with:.5f} seconds")
````
