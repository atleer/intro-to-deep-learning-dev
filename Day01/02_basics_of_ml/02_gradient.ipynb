## Section 1: ML is Iterative Guessing.  Iteration can be done with For Loops

When training a neural network, you're making the model guess what the output should be over and over again. Between each guess, the model parameters are updated until the guess is good enough. In other words, it's an iterative improvement of the model. This iteration can be programmed with a for loop. A for loop lets us perform an operation a predefined number of times and is an incredibly useful programming concept. In this section, you will get som practice writing for loops.

| Code | Description |
| --- | --- |
| `for letter in ["a", "b", "c"]:` <br> &nbsp;&nbsp;&nbsp;&nbsp; `print(letter)` | Print every `letter` in the list `["a", "b", "c"]`|
| `for i in range(5):` <br> &nbsp;&nbsp;&nbsp;&nbsp; `print(i)` | Print every number `i` between 0 and 5 (not including 5)|
| `for i in range(2, 7):` <br> &nbsp;&nbsp;&nbsp;&nbsp; `print(i)` | Print every number `i` between 2 and 7 (not including 7)|
| `for i in range(1, 10, 2):` <br> &nbsp;&nbsp;&nbsp;&nbsp; `print(i)` | Print every **second** number `i` between 1 and 10 (not including 10)|
| `for x_i in x:` <br> &nbsp;&nbsp;&nbsp;&nbsp; `print(x_i)` | Print every element `x_i` in the collection `x`|
| `for x_i in x:` <br> &nbsp;&nbsp;&nbsp;&nbsp; `for y_i in y`: <br> &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; `print(x_i, y_i)` | Print every element `x_i` and `y_i` in the collections `x` and `y`|

Let's say that we have recorded some data and we know that this data should be described by a function of the form:

$$ y = a * x$$

However, we don't know what `a`. We could of course use one of the standard linear regression functions that many python libraries offer to do this, but let's do it with machine learning instead. And we'll do it manually to get a better intuition for what is actually going on under the hood.

The observed value is `y_obs = 10.0` for a given `x = 2.0`. These are defined in the code cell below. As mentioned in section 1, in machine learning, we start with a guess, so let's set `a = 1.0` and see how close the predicted value `y_pred` given this `a` is. Run the cell below.

````python
# Data
x = 2.0
y_obs = 10.0

# Parameter
a = 1.0 # initial guess

y_pred = a * x
print(f"Predicted y-value: {y_pred}, intial guess for a: {a}")
````

The predicted value `y_pred` wasn't very close to our observed data point `y_pred`. To do something useful with the deviation between the observed and predicted y-value, we need to quantify their difference.

**Exercise**: Calculate the loss between the predicted and observed y-values. Here, the loss can be calculated as the squared deviation between the two numbers.

````python
(y_pred - y_obs)**2
````

Now that we have a way to quantify how far we were off with our initial guess, we can use that information to get closer to the observed y-value with our next guess. That's where the gradient comes in. The gradient tells us how much the loss will change if we change the parameter `a`. In mathematical terms, it's the derivative of the loss with respect to the gradient.

$$grad_a = \frac{\Delta loss}{\Delta a}$$

$$grad_a = \frac{\partial{loss}}{\partial{a}}$$

with $y_{pred} = a * x$ and $loss = y_{obs}-y_{pred}$, this becomes 

$$ grad_a = \frac{\partial{(y_{obs}-a * x)}}{\partial{a}}$$

$$ grad_a = -2(y_{obs}-a*x)*x$$

$$ grad_a = -2(y_{obs}-y_{pred})*x$$

**Exercise**: Write the code for the last equation above.


````python
grad_a = -2*(y_obs-y_pred)*x
grad_a
````

The as mentioned, the gradient tells you how much the loss will change given a change in `a`. We'll now use the gradient we calculated to update the parameter `a` before our next guess.

**Exercise**: Calculate a new parameter value `a_new` using the current value of `a` and the gradient calculated above.

````python
a_new = -grad_a*a
a_new
````

Now we're ready to make our next guess. 

**Exercise**: Calculate `y_pred` again but use the new parameter value `a_new` instead of the original `a`.

**Hint**: This guess will actually be worse, so don't worry, that's to be expected.

````python
y_pred = a_new * x
y_pred
````

There is another important *hyperparameter* (different from a regular parameter) called the **learning rate**, which determines how we learn in machine learning. Without an appropriate learning rate, the updated parameter values won't produce better guesses. The learning rate is multiplied with the gradient and the current parameter value.

**Exercise**: Calculate the updated parameter value `a_new` again, but this time, multiply it with the `learning_rate` defined in the cell below.

````python
learning_rate = 0.1
a_new = -learning_rate*grad_a*a
a_new
````

**Exercise**: Calculate `y_pred` with the `a_new` that we have now. Are we closer to or further away from the observed `y_obs` this time?

````python
y_pred = a_new*x
y_pred
````

**Exercise**: The code below contains all the steps above that go into updating the parameter. Put the steps in a for loop that stops after 5 iterations.

````python
# Forward pass
y_pred = a * x

# calculate loss
loss = (y_pred - y_obs)**2

print(f"a={a:.3f}, y_pred={y_pred:.2f}, loss={loss:.3f}")

# Compute gradient
grad_a = 2 * (y_pred - y_obs) * x

# Update the parameter
a = a - learning_rate * grad_a
````

````python
for step in range(5):
    # Forward pass
    y_pred = a * x

    # calculate loss
    loss = (y_pred - y_obs)**2

    print(f"a={a:.3f}, y_pred={y_pred:.2f}, loss={loss:.3f}")

    # Compute gradient
    grad_a = 2 * (y_pred - y_obs) * x

    # Update the parameter
    a = a - learning_rate * grad_a
````

## Section 3: `torch.tensor()` has Autograd directly implmented, so it can compute deriviates. 

The great thing about PyTorch is that it takes care of the gradient calculation and parameter updates for you. This is particularly useful when you have a big model with a lot of parameters and layers that need to be kept track of and updated. 

```python
x = torch.tensor([2.], requires_grad=True)
y = x ** 2
y.backward()
x.grad  # tensor([7.])
```

| `x.detach()` | Stop gradient tracking |
| `with x.no_grad():` | Pause gradient tracking temporarily|
| `x.zero_()` | Reset the gradient. |

````python
import torch

# Data
x = torch.tensor(2.0)         # input (no gradient needed)
y_obs = torch.tensor(10.0)    # target

# Parameter to learn
a = torch.tensor(1.0, requires_grad=True)

lr = 0.1

for step in range(6):
    # Forward pass
    y_pred = a * x
    loss = (y_pred - y_obs)**2

    print(f"step {step}: a={a.item():.3f}, y_pred={y_pred.item():.2f}, loss={loss.item():.3f}")

    # Backward pass — autograd computes dl/da for us
    loss.backward()

    # Update parameter
    with torch.no_grad():
        a -= lr * a.grad

    # Reset gradient for next step
    a.grad.zero_()
````
