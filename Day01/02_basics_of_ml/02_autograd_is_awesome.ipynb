## Section 1: ML is Iterative Guessing.  Iteration can be done with For-Loops


When training a neural network, we would like to go through many iterations of passing the input to the network, comparing the network's output to the true values from the dataset, and updating the parameters of the network. To do that, we use a for loop, which allows us to perfor the same set of operations a predefined number of times.

| Code | Description |
| --- | --- |
| `for letter in ["a", "b", "c"]:` <br> &nbsp;&nbsp;&nbsp;&nbsp; `print(letter)` | Print every `letter` in the list `["a", "b", "c"]`|
| `for i in range(5):` <br> &nbsp;&nbsp;&nbsp;&nbsp; `print(i)` | Print every number `i` between 0 and 5 (not including 5)|
| `for i in range(2, 7):` <br> &nbsp;&nbsp;&nbsp;&nbsp; `print(i)` | Print every number `i` between 2 and 7 (not including 7)|
| `for i in range(1, 10, 2):` <br> &nbsp;&nbsp;&nbsp;&nbsp; `print(i)` | Print every **second** number `i` between 1 and 10 (not including 10)|
| `for x_i in x:` <br> &nbsp;&nbsp;&nbsp;&nbsp; `print(x_i)` | Print every element `x_i` in the collection `x`|
| `for x_i in x:` <br> &nbsp;&nbsp;&nbsp;&nbsp; `for y_i in y`: <br> &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; `print(x_i, y_i)` | Print every element `x_i` and `y_i` in the collections `x` and `y`|
| `new_list = []` or `new_list = list()`| Initialize an empty list named `new_list` |
| `new_list.append(element)`| Append an element to the list `new_list` |

## Section 2: ML involves adjusting guesses based on error. 

The output of the network is a score for each class that reflects the networks belief that a data point corresponds to a given class. This is not the same as a class label. To get the output in the form of a class guess by the network, we have to determine which class the network thinks that a data point is most likely to belong to using the output scores.

| Code | Description |
| --- | --- |
| `torch.argmax(x, dim=1)` | Finds the index at which the tensor `x` has the highest value along its first dimension. |
| `a == b` | Compares each element of the tensors (or arrays) `a` and `b` and returns a tensor of boolean (`True` or `False`) values indicating which elements are the same (`True`) and different (`False`). |
| `(a == b).sum()` or `torch.sum(a == b)`| Sums all elements where the condition `a == b` is `True`. `True` is counted as 1, `False` as 0. |
| `(a == b).sum() / len(a)` | Calculates proportion of elements in `a` where the element is equal to the corresponding element in `b`.|

Run the cell below to create a network, select a subset of the data, and pass this subset through the network to get an output `y_net`.

````python
# Manual gradient descent, no autograd

# Data
x = 2.0
y_obs = 10.0

# Parameter
a = 1.0
learning_rate = 0.1

for step in range(6):
    # Forward pass
    y_pred = a * x
    loss = (y_pred - y_obs)**2

    print(f"step {step}: a={a:.3f}, y_pred={y_pred:.2f}, loss={loss:.3f}")

    # Compute gradient
    grad_a = 2 * (y_pred - y_obs) * x

    # Update
    a = a - learning_rate * grad_a
````

| Code | Description |
| :-- | :-- |
| `x.requires_grad` | Is gradient tracking on? |
| `x.detach()` | Stop gradient tracking |
| `with x.no_grad():` | Pause gradient tracking temporarily|
| `x.zero_()` | Reset the gradient. |



````python
import torch

x = torch.tensor([2.], requires_grad=True)
y = x * 5 + 1
y.backward()
x.grad  
````

````python
import torch

# Data
x = torch.tensor(2.0)         # input (no gradient needed)
y_obs = torch.tensor(10.0)    # target

# Parameter to learn
a = torch.tensor(1.0, requires_grad=True)

lr = 0.1

for step in range(6):
    # Forward pass
    y_pred = a * x
    loss = (y_pred - y_obs)**2

    print(f"step {step}: a={a.item():.3f}, y_pred={y_pred.item():.2f}, loss={loss.item():.3f}")

    # Backward pass — autograd computes dl/da for us
    loss.backward()

    # Update parameter
    with torch.no_grad():
        a -= lr * a.grad

    # Reset gradient for next step
    a.grad.zero_()
````
