```python
from abc import ABC, abstractmethod

class IGreet(ABC):

    @abstractmethod
    def get_greeting(self) -> str: ...
    
    @abstractmethod
    def greet(self, name) -> str: ...


class BaseGreet(IGreet):
    
    def log(self):
        print('Preparing Greeting...')

    def greet(self, name):
        self.log()
        greeting = self.get_greeting()
        return f"{greeting}, {name}"
    

class EnglishGreet(BaseGreet):

    def get_greeting(self) -> str:
        return "Hello"

class GermanGreet(BaseGreet):

    def get_greeting(self) -> str:
        return 'Hallo'
    



# EnglishGreet().greet('Atle')
GermanGreet().greet('Atle')
```
