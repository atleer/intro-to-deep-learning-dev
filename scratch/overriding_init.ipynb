```python
from abc import ABC, abstractmethod

class IDataset(ABC):

    @abstractmethod
    def get_data(self): ...
        
    @abstractmethod
    def save(self, filename): ...



class BaseDataset(IDataset):

    def __init__(self, *args, **kwargs) -> None:
        if kwargs.get('verbose'):
            print('init')

    def save(self, filename):
        print(f"saved to {filename}: {self.get_data()}")



class Dataset(BaseDataset):
    
    def get_data(self):
        return [1, 2, 3]


class MyDataset(BaseDataset):

    def get_data(self):
        return [4, 5, 6]


class MyDataset2(BaseDataset):

    def __init__(self, *args, **kwargs) -> None:
        super().__init__(*args, **kwargs)
        self.data = [7, 8, 9]
        self.layer1 = [2, 3, 4]

    

from functools import lru_cache, cached_property

class MyDataset3(BaseDataset):

    @property
    @lru_cache
    def _data(self):
        print('generating...')
        return [10, 11, 12]

    def get_data(self):
        return self._data


class PostdocCompatiblMyDataset2(MyDataset2):

    def solve(self):
        self.save('ff1.nd')

# MyDataset(data=[6, 7, 8], verbose=True).save('data')
# MyDataset2().save('myfile.csv')
MyDataset3().save('ff2.csv')
MyDataset3().save('ff2.csv')
MyDataset3().save('ff2.csv')
```
